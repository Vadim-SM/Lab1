## 3. Завантаження набору даних
У роботі використовується навчальний набір Iris із бібліотеки scikit-learn. Він містить вимірювання квіток трьох видів ірису.

```python
iris = load_iris(as_frame=True)
print(type(iris))
print(iris.keys())
```

```python
print("Назви ознак:")
for feature_name in iris.feature_names:
    print("-", feature_name)

print("\nНазви класів:")
for target_name in iris.target_names:
    print("-", target_name)
```

```python
df = iris.frame.copy()
display(df.head())
```

```python
print(df["target"].value_counts().sort_index())

target_mapping = {
    0: "setosa",
    1: "versicolor",
    2: "virginica"
}
print(target_mapping)

df["species"] = df["target"].map(target_mapping)
df = df.drop(columns=["target"])
```

```python
df = df.rename(
    columns={
        "sepal length (cm)": "sepal_length_cm",
        "sepal width (cm)": "sepal_width_cm",
        "petal length (cm)": "petal_length_cm",
        "petal width (cm)": "petal_width_cm"
    }
)
display(df.head())
```

## 6. Візуалізація даних
Гістограми використовуються для первинного дослідження розподілів числових ознак.

```python
feature_columns = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]

data[feature_columns].hist(
    bins=15,
    figsize=(10, 8)
)
plt.suptitle("Розподіли числових ознак набору Iris")
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

**Відповіді на завдання для аналізу:**
1. Найменші значення має ознака `petal_width_cm` (ширина пелюстки).
2. Найбільший діапазон значень має ознака `petal_length_cm` (довжина пелюстки).
3. Ні, розподіли мають різну форму (наприклад, розподіл `sepal_width_cm` найбільш наближений до нормального/дзвіноподібного, тоді як ознаки пелюстки мають сильні розриви).
4. Кілька груп значень чітко помітно для ознак `petal_length_cm` та `petal_width_cm` (яскраво виражена ліва група).

```python
class_counts = data["species"].value_counts().sort_index()

class_counts.plot(
    kind="bar",
    figsize=(8, 5)
)
plt.title("Кількість об’єктів кожного класу")
plt.xlabel("Вид ірису")
plt.ylabel("Кількість об’єктів")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Питання: Чи є набір даних збалансованим?**
Так, набір даних є абсолютно збалансованим. Кожен із трьох класів містить рівно по 50 об'єктів (приблизно 33.33% від загальної кількості).

```python
for species_name, group in data.groupby("species"):
    plt.scatter(
        group["petal_length_cm"],
        group["petal_width_cm"],
        label=species_name,
        alpha=0.8
    )
plt.title("Залежність довжини та ширини пелюстки")
plt.xlabel("Довжина пелюстки, см")
plt.ylabel("Ширина пелюстки, см")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

**Відповіді на питання для аналізу:**
1. **Чи утворюють об’єкти різних класів окремі групи?** Так, об’єкти утворюють помітні кластери (групи) відповідно до свого виду.
2. **Який клас відокремлюється найкраще?** Найкраще відокремлюється клас `setosa` (знаходиться окремо у нижньому лівому куті).
3. **Які два класи частково перекриваються?** Частково перекриваються класи `versicolor` та `virginica`.
4. **Чи можна використати ці ознаки для розпізнавання виду квітки?** Так, ознаки довжини та ширини пелюстки відмінно підходять для класифікації ірисів.

## 7. Постановка задачі машинного навчання

### Об’єкт дослідження
Одним об’єктом є одна квітка ірису, для якої виміряно характеристики чашолистка та пелюстки.

### Ознаки
Використовуються чотири числові ознаки:
1. довжина чашолистка;
2. ширина чашолистка;
3. довжина пелюстки;
4. ширина пелюстки.

### Цільова змінна
Цільовою змінною є `species` – вид квітки ірису.

### Тип задачі
Задача належить до навчання з учителем і є задачею багатокласової класифікації.

### Формулювання задачі
Необхідно за чотирма числовими характеристиками квітки автоматично визначити, до якого з трьох видів ірису вона належить: setosa, versicolor або virginica.

```python
X = data.drop(columns=["species"])
y = data["species"]

print("Розмір матриці ознак X:", X.shape)
print("Розмір цільової змінної y:", y.shape)

display(X.head())
display(y.head())
```
